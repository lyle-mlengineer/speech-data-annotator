```python

import librosa
import soundfile as sf
import os
```

```python
MAX_AUDIO_DURATION_SECONDS = 30
MIN_AUDIO_DURATION_SECONDS = 10
SAMPLING_RATE = 16000
```

```python
audio_path = "C:/Datasets/audio/maongezi/raw/GCsmZA08oD8/GCsmZA08oD8.mp3"
```

```python
audio, sr = librosa.load(audio_path)
```

```python
audio_duration = round(len(audio) / sr, 2)
print(f"The audio is {audio_duration} seconds long")
for i, start_idx in enumerate(range(0, len(audio), MAX_AUDIO_DURATION_SECONDS * sr)):
    end_idx = start_idx + MAX_AUDIO_DURATION_SECONDS * sr
    audio_segment = audio[start_idx:end_idx]
    if len(audio_segment) < MIN_AUDIO_DURATION_SECONDS * sr:
        continue
    segment_duration = round(len(audio_segment) / sr, 2)
    print(f"Audio segment {i+1} is {segment_duration} seconds long")
    # sf.write(os.path.join(".", f"{i+1}.wav"), audio_segment, sr)
```
